```python
import ipywidgets as widgets
from IPython.display import display, clear_output
import re

# --- 1. 定义数据 ---
original_file_names = [
    "CHC101_20251203_AMB_20251203000001_Passed_20251203.xlsx",
    "CHC101_20251203_AMB_20251204000001_Passed_20251204.xlsx",
    "CHC101_20251205_AMB_20251205000001_Passed_20251205.xlsx"
]

# --- 2. 创建下拉菜单控件 ---
dropdown = widgets.Dropdown(
    options=original_file_names,
    value=original_file_names[0],
    description='选择文件:',
    style={'description_width': 'initial'}
)

# --- 3. 定义点击按钮后要执行的逻辑 ---
button = widgets.Button(description="运行处理逻辑", button_style='success')
output = widgets.Output()

def on_button_clicked(b):
    with output:
        clear_output() # 每次点击先清空旧的日志
        selected_file = dropdown.value
        
        # 核心逻辑：截取 tx_dt (最后8位)
        match = re.search(r'(\d{8})\.xlsx$', selected_file)
        tx_dt = match.group(1) if match else "Unknown"
        
        print(f"📂 当前处理: {selected_file}")
        print(f"📅 提取日期: {tx_dt}")
        
        # 这里放置你的 Spark SQL 或 Iceberg 写入逻辑
        # spark.sql(f"SELECT * FROM ... WHERE tx_dt = '{tx_dt}'")
        print(f"✅ 执行完成！数据已同步至 Iceberg 分区: {tx_dt}")

button.on_click(on_button_clicked)

# --- 4. 显示界面 ---
display(dropdown, button, output)
```

```python
import ipywidgets as widgets
from IPython.display import display
import re

# 1. 定义你的原始数据列表
original_file_names = [
    "CHC101_20251203_AMB_20251203000001_Passed_20251203.xlsx",
    "CHC101_20251203_AMB_20251204000001_Passed_20251204.xlsx",
    "CHC101_20251205_AMB_20251205000001_Passed_20251205.xlsx"
]

# 2. 初始化全局变量（方便后续 Cell 直接引用）
# 这里我们先给它们赋初始值
original_file_name = original_file_names[0]
tx_dt = re.search(r'(\d{8})\.xlsx$', original_file_name).group(1)

# 3. 创建下拉菜单控件
dropdown = widgets.Dropdown(
    options=original_file_names,
    value=original_file_name,
    description='选择文件:',
    disabled=False,
    style={'description_width': 'initial'},
    layout={'width': 'max-content'}
)

# 4. 定义回调函数：当下拉菜单值改变时，自动更新全局变量
def on_change(change):
    if change['type'] == 'change' and change['name'] == 'value':
        global original_file_name, tx_dt
        # 更新文件名变量
        original_file_name = change['new']
        # 自动提取最后8位日期
        match = re.search(r'(\d{8})\.xlsx$', original_file_name)
        tx_dt = match.group(1) if match else "Unknown"
        
        print(f"✅ 变量已同步更新: tx_dt='{tx_dt}'")

# 绑定观察事件
dropdown.observe(on_change)

# 显示控件
display(dropdown)
```
